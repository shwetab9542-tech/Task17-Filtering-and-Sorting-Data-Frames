```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
#Filtering Example 1: Passengers who survived
survived = df[df["Survived"] == 1]
survived.head()
```

```python
#Filtering Example 2: Passengers who did not survive
not_survived = df[df["Survived"] == 0]
not_survived.head()
```

```python
#Filtering Example 3: Passengers older than 30
older_than_30 = df[df["Age"] > 30]
older_than_30.head()
```

```python
#Filtering Example 4: Female passengers
female_passengers = df[df["Sex"] == "female"]
female_passengers.head()
```

```python
#Filtering Example 5: First-class passengers
first_class = df[df["Pclass"] == 1]
first_class.head()
```

```python
#Multiple conditions
female_survivors = df[(df["Sex"] == "female") & (df["Survived"] == 1)]
female_survivors.head()
```

```python
#Sorting
df.sort_values("Age", ascending=True).head(10)
```

```python
#Sorting by multiple columns
df.sort_values(
    by=["Pclass", "Fare"],
    ascending=[True, False]
).head(10)
```

```python
#Short analysis
print("Total passengers:", len(df))
print("Survived passengers:", df["Survived"].sum())
print("Female passengers:", len(female_passengers))
print("First-class passengers:", len(first_class))
```
